```python
# Combine the closing price by dates
# Combine closing prices for the nine stocks and the NIFTY 50 benchmark
benchmark = "NIFTY50"
stocks = [name for name in datasets if name != benchmark]

prices = pd.DataFrame()

for name in stocks + [benchmark]:
    df = pd.read_csv(
        folder / f"{name}_5year.csv",
        usecols=["Date", "Close"],
        parse_dates=["Date"]
    )
    prices[name] = df.set_index("Date")["Close"]

prices = prices.sort_index()

assert len(stocks) == 9
assert prices.index.is_unique
assert prices.notna().all().all()
assert prices.gt(0).all().all()

print("Stocks used for optimization:", stocks)
print("Number of stocks:", len(stocks))
print("Benchmark:", benchmark)
print("Price table shape:", prices.shape)

display(prices.head(3))
display(prices.tail(3))
```

```python
# Restore the observed trading-date sequence before calculating returns.
# A missing session then produces a missing return instead of a multi-day return.
daily_returns = (
    prices.reindex(observed_dates)
    .pct_change(fill_method=None)
    .dropna()
)

print("Daily return table shape:", daily_returns.shape)
display(daily_returns.head(3).style.format("{:.3%}"))
display(daily_returns.tail(3).style.format("{:.3%}"))
```

```python
# Average daily return and risk
daily_summary = pd.DataFrame({
    "Mean daily return": daily_returns.mean(),
    "Daily variance": daily_returns.var(),
    "Daily volatility": daily_returns.std()
})

display(daily_summary.style.format({
    "Mean daily return": "{:.4%}",
    "Daily variance": "{:.6f}",
    "Daily volatility": "{:.3%}"
}))
```

```python
# Annualize returns and risk
trading_days = 252
risk_free_rate = 0.05

annual_returns = daily_returns[stocks].mean() * trading_days
annual_cov = daily_cov * trading_days
annual_volatility = np.sqrt(np.diag(annual_cov))

annual_summary = pd.DataFrame({
    "Annualized mean return": annual_returns,
    "Annualized volatility": annual_volatility
})

display(annual_summary.style.format("{:.2%}"))
```

```python
# Minimum variance Portfolio
from scipy.optimize import minimize

n = len(stocks)
bounds = [(0, 1)] * n
fully_invested = {"type": "eq", "fun": lambda w: w.sum() - 1}

def portfolio_variance(w):
    return w @ annual_cov @ w

result = minimize(
    portfolio_variance,
    x0=np.full(n, 1 / n),
    method="SLSQP",
    bounds=bounds,
    constraints=fully_invested,
    options={"ftol": 1e-12}
)

if not result.success:
    raise RuntimeError(result.message)

min_var_weights = pd.Series(result.x, index=stocks)
min_var_return = min_var_weights @ annual_returns
min_var_volatility = np.sqrt(result.fun)

print(min_var_weights.to_string(float_format="{:.2%}".format))
print(f"Annualized return:     {min_var_return:.2%}")
print(f"Annualized volatility: {min_var_volatility:.2%}")
```

```python
from matplotlib.ticker import PercentFormatter

target_returns = np.linspace(min_var_return, annual_returns.max(), 50)
frontier_volatility = []
frontier_allocations = []
weights = min_var_weights.to_numpy()

for target in target_returns:
    target_constraint = {
        "type": "eq",
        "fun": lambda w, target=target: w @ annual_returns - target
    }

    solution = minimize(
        portfolio_variance,
        x0=weights,
        method="SLSQP",
        bounds=bounds,
        constraints=[fully_invested, target_constraint],
        options={"ftol": 1e-12}
    )

    if not solution.success:
        raise RuntimeError(f"Target {target:.2%}: {solution.message}")

    weights = solution.x
    frontier_volatility.append(np.sqrt(solution.fun))
    frontier_allocations.append(weights.copy())

fig, ax = plt.subplots()
ax.plot(frontier_volatility, target_returns, label="Efficient frontier")
ax.scatter(
    min_var_volatility, min_var_return,
    color="red", label="Minimum-variance portfolio", zorder=3
)
ax.set(
    xlabel="Annualized volatility",
    ylabel="Annualized mean return",
    title="Efficient frontier of the nine stocks"
)
ax.xaxis.set_major_formatter(PercentFormatter(1))
ax.yaxis.set_major_formatter(PercentFormatter(1))
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# how allocations change with target return
allocation_table = pd.DataFrame(
    frontier_allocations,
    index=pd.Index(target_returns, name="Target annual return"),
    columns=stocks
)

# Display the first, middle, and highest target-return portfolios.
selected_allocations = allocation_table.iloc[[0, len(allocation_table) // 2, -1]]

display(
    selected_allocations.style
    .format("{:.2%}")
    .format_index("{:.2%}", axis=0)
)
```

```python
# maximim sharp portfolio finding
def negative_sharpe(w):
    excess_return = w @ annual_returns - risk_free_rate
    volatility = np.sqrt(portfolio_variance(w))
    return -excess_return / volatility

optimal = minimize(
    negative_sharpe,
    x0=np.full(n, 1 / n),
    method="SLSQP",
    bounds=bounds,
    constraints=fully_invested,
    options={"ftol": 1e-12}
)

if not optimal.success:
    raise RuntimeError(optimal.message)

optimal_weights = pd.Series(optimal.x, index=stocks)
optimal_return = optimal_weights @ annual_returns
optimal_volatility = np.sqrt(portfolio_variance(optimal.x))
optimal_sharpe = (optimal_return - risk_free_rate) / optimal_volatility

print(optimal_weights.to_string(float_format="{:.2%}".format))
print(f"Annualized mean return: {optimal_return:.2%}")
print(f"Annualized volatility:  {optimal_volatility:.2%}")
print(f"Sharpe ratio:           {optimal_sharpe:.3f}")
```

```python
equal_weights = pd.Series(1 / n, index=stocks)

portfolio_weights = pd.DataFrame({
    "Equal weight": equal_weights,
    "Minimum variance": min_var_weights,
    "Maximum Sharpe": optimal_weights
})

print("Allocation across the nine stocks")
display(portfolio_weights.style.format("{:.2%}"))

comparison_returns = daily_returns[stocks] @ portfolio_weights
comparison_returns["NIFTY 50"] = daily_returns[benchmark]

mean_return = comparison_returns.mean() * trading_days
volatility = comparison_returns.std() * np.sqrt(trading_days)

comparison = pd.DataFrame({
    "Annualized mean return": mean_return,
    "Annualized volatility": volatility,
    "Sharpe ratio": (mean_return - risk_free_rate) / volatility
})

print("Five-year in-sample performance")
display(comparison.style.format({
    "Annualized mean return": "{:.2%}",
    "Annualized volatility": "{:.2%}",
    "Sharpe ratio": "{:.3f}"
}))
```

```python
# test sensitivity to the estimation period
training_end = start_date + pd.DateOffset(years=3)
training_returns = daily_returns.loc[:training_end, stocks]

training_mean = training_returns.mean() * trading_days
training_cov = training_returns.cov() * trading_days

def training_variance(w):
    return w @ training_cov @ w

def training_negative_sharpe(w):
    excess_return = w @ training_mean - risk_free_rate
    return -excess_return / np.sqrt(training_variance(w))

training_weights = pd.DataFrame(index=stocks)
training_weights["Equal weight"] = equal_weights

for name, objective in {
    "Minimum variance": training_variance,
    "Maximum Sharpe": training_negative_sharpe
}.items():
    fit = minimize(
        objective,
        x0=equal_weights.to_numpy(),
        method="SLSQP",
        bounds=bounds,
        constraints=fully_invested,
        options={"ftol": 1e-12}
    )
    if not fit.success:
        raise RuntimeError(f"{name}: {fit.message}")

    training_weights[name] = fit.x

optimized_names = ["Minimum variance", "Maximum Sharpe"]
weight_change = (
    portfolio_weights[optimized_names] - training_weights[optimized_names]
) * 100

print("Weight change from three-year to five-year estimates (percentage points)")
display(weight_change.round(2))
```

```python
# Evaluate the final 2 year
test_returns = daily_returns.loc[daily_returns.index > training_end]

test_portfolios = test_returns[stocks] @ training_weights
test_portfolios["NIFTY 50"] = test_returns[benchmark]

test_mean = test_portfolios.mean() * trading_days
test_volatility = test_portfolios.std() * np.sqrt(trading_days)

test_comparison = pd.DataFrame({
    "Annualized mean return": test_mean,
    "Annualized volatility": test_volatility,
    "Sharpe ratio": (test_mean - risk_free_rate) / test_volatility
})

print("Final two years: out-of-sample performance")
display(test_comparison.style.format({
    "Annualized mean return": "{:.2%}",
    "Annualized volatility": "{:.2%}",
    "Sharpe ratio": "{:.3f}"
}))
```
